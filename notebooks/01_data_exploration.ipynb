```python
from datasets import load_dataset

dataset = load_dataset(
    "bitext/Bitext-retail-ecommerce-llm-chatbot-training-dataset"
)
```

```python
# all unique 'intent' values in the dataset
unique_intents = dataset["train"].unique("intent")
print(f"Number of unique intents: {len(unique_intents)}")
print(unique_intents)
```

```python
import pandas as pd

df = dataset["train"].to_pandas()

mapping = (
    df[["intent", "category"]]
    .drop_duplicates()
    .sort_values(["category", "intent"])
)

print(mapping.to_string(index=False))
```

```python
print(df.shape)
print(df.columns.tolist())

print(
    df[["instruction", "intent", "response"]]
    .sample(10, random_state=42)
    .to_string(index=False)
)
```

```python
df.groupby("intent")["category"].nunique().sort_values(ascending=False)
```

```python
from collections import defaultdict

intent_categories = defaultdict(set)

for row in dataset["train"]:
    intent_categories[row["intent"]].add(row["category"])

for intent in sorted(intent_categories):
    print(f"{intent:30} -> {intent_categories[intent]}")
```

```python
# all unique 'category' values in the dataset
unique_categories = dataset["train"].unique("category")
print(f"Number of unique categories: {len(unique_categories)}")
print(unique_categories)
```

```python
#how many intents are there per category
category_intent_counts = df.groupby("category")["intent"].nunique().sort_values(ascending=False)
category_intent_counts
```

```python
#all unique 'tags' values in the dataset
unique_tags = dataset["train"].unique("tags")
print(f"Number of unique tags: {len(unique_tags)}")
print(unique_tags)
```

```python
#Save the dataset to a ../data/raw parquet file
dataset["train"].to_parquet("../data/raw/Bitext-retail-ecommerce-llm-chatbot-training-dataset.parquet")
```

```python
intent_map = (
    dataset["train"]
    .to_pandas()
    .groupby("category")["intent"]
    .unique()
)

for category, intents in intent_map.items():
    print(f"\n{category} ({len(intents)})")
    for intent in sorted(intents):
        print(f"  - {intent}")
```
